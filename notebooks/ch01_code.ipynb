## 1.4　関数 ― 処理をまとめる

````python
def dice_score(pred, target):    """予測マスクと正解マスクのDice係数を計算する"""    intersection = (pred * target).sum()    return 2.0 * intersection / (pred.sum() + target.sum() + 1e-8)# pred・target は予測マスクと正解マスク（NumPy配列。第2章で学びます）。呼び出しの形は次のとおり：# score = dice_score(pred, target)# print(f"Dice = {score:.3f}")
````
